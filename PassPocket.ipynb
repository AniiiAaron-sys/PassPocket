```python
import secrets
import string

vault = []

LOWERCASE = string.ascii_lowercase
UPPERCASE = string.ascii_uppercase
NUMBERS = string.digits
SYMBOLS = "!@#$%^&*()-_=+"

print("PassPocket is ready!")
```

```python
def generate_password(length=16, use_upper=True,
                      use_lower=True, use_numbers=True,
                      use_symbols=True):

    # Building the character groups
    groups = []

    if use_lower:
        groups.append(LOWERCASE)
    if use_upper:
        groups.append(UPPERCASE)
    if use_numbers:
        groups.append(NUMBERS)
    if use_symbols:
        groups.append(SYMBOLS)

    # Check
    if not groups:
        raise ValueError("Select at least one character type.")

    if not 4 <= length <= 128:
        raise ValueError("Length must be between 4 and 128.")

    # Pick at least one character from each selected group
    password_chars = [
        secrets.choice(group) for group in groups
    ]

    # Fill the remaining
    alphabet = "".join(groups)

    while len(password_chars) < length:
        password_chars.append(secrets.choice(alphabet))

    # Shuffle so required character types aren't predictable
    secrets.SystemRandom().shuffle(password_chars)

    return "".join(password_chars)


# Test
test_password = generate_password(length=20)
print("Generated password:", test_password)
print("Password length:", len(test_password))
```

```python
def password_creator():
    print("\n--- Password Creator ---")

    try:
        length = int(input("Password length (4-128): "))

        upper = input("Include uppercase letters? (y/n): ").lower()
        lower = input("Include lowercase letters? (y/n): ").lower()
        numbers = input("Include numbers? (y/n): ").lower()
        symbols = input("Include symbols? (y/n): ").lower()

        choices = [upper, lower, numbers, symbols]

        if any(choice not in ("y", "n") for choice in choices):
            print("Please answer each question with y or n.")
            return

        password = generate_password(
            length=length,
            use_upper=(upper == "y"),
            use_lower=(lower == "y"),
            use_numbers=(numbers == "y"),
            use_symbols=(symbols == "y")
        )

        print("\nYour generated password:")
        print(password)

    except ValueError as error:
        print("Error:", error)


password_creator()
```

```python
# add login


def add_entry(site, username, password):
    site = site.strip()
    username = username.strip()

    if not site or not username or not password:
        print("All fields are required.")
        return

    entry = {
        "site": site,
        "username": username,
        "password": password
    }

    vault.append(entry)
    print(f"Added login for {site}.")


def list_entries():
    print("\n--- Saved Logins ---")

    if not vault:
        print("Your vault is empty.")
        return

    for number, entry in enumerate(vault, start=1):
        print(f"{number}. {entry['site']} — {entry['username']}")


def search_entries(query):
    query = query.strip().lower()

    if not query:
        print("Enter a website or app name to search.")
        return []

    matches = [
        entry for entry in vault
        if query in entry["site"].lower()
    ]

    if not matches:
        print("No matching logins found.")
        return []

    for entry in matches:
        print("\nWebsite:", entry["site"])
        print("Username:", entry["username"])
        print("Password:", entry["password"])

    return matches


def delete_entry(number):
    if not 1 <= number <= len(vault):
        print("Invalid entry number.")
        return

    entry = vault[number - 1]
    confirmation = input(
        f"Delete the login for {entry['site']}? (y/n): "
    ).lower()

    if confirmation == "y":
        vault.pop(number - 1)
        print("Login deleted.")
    else:
        print("Deletion cancelled.")
```

```python
# main menu

# Add two sample logins
add_entry("example.com", "student123", "DemoPass!123")
add_entry("games.example", "player456", "FakeGame!456")

# Show the entries
list_entries()

# Search for a website
search_entries("example.com")
```

```python
def main_menu():
    while True:
        print("\n" + "=" * 30)
        print("       PassPocket")
        print("=" * 30)
        print("1. Generate a password")
        print("2. Add a login")
        print("3. List saved logins")
        print("4. Search for a login")
        print("5. Delete a login")
        print("6. Exit")

        choice = input("\nChoose an option (1-6): ").strip()

        if choice == "1":
            password_creator()

        elif choice == "2":
            site = input("Website or app: ")
            username = input("Username: ")
            password = input("Password (or paste a generated one): ")
            add_entry(site, username, password)

        elif choice == "3":
            list_entries()

        elif choice == "4":
            query = input("Website or app to search for: ")
            search_entries(query)

        elif choice == "5":
            list_entries()

            if vault:
                try:
                    number = int(
                        input("Entry number to delete: ")
                    )
                    delete_entry(number)
                except ValueError:
                    print("Please enter a valid number.")

        elif choice == "6":
            print("Thanks for trying PassPocket!")
            break

        else:
            print("Invalid choice. Please choose 1-6.")


# Run PassPocket
main_menu()
def main_menu():
    while True:
        print("\n" + "=" * 30)
        print("         PassPocket")
        print("=" * 30)
        print("1. Generate a password")
        print("2. Add a login")
        print("3. List saved logins")
        print("4. Search for a login")
        print("5. Delete a login")
        print("6. Exit")

        choice = input("\nChoose an option (1-6): ").strip()

        if choice == "1":
            password_creator()

        elif choice == "2":
            site = input("Website or app: ")
            username = input("Username: ")
            password = input("Password (or paste a generated one): ")
            add_entry(site, username, password)

        elif choice == "3":
            list_entries()

        elif choice == "4":
            query = input("Website or app to search for: ")
            search_entries(query)

        elif choice == "5":
            list_entries()

            if vault:
                try:
                    number = int(input("Entry number to delete: "))
                    delete_entry(number)
                except ValueError:
                    print("Please enter a valid number.")
            else:
                print("Your vault is empty.")

        elif choice == "6":
            print("Thanks for trying PassPocket!")
            break

        else:
            print("Invalid choice. Please choose 1-6.")


# Run PassPocket
main_menu()
```
